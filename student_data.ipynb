Analyzing Data


```python
import pandas as pd 
import numpy as np
pd.options.display.max_rows = 9999
data = pd.read_csv('./data/data.csv')
print(data)
```

```python
drop_data = data.dropna()
print(drop_data)
drop_data.corr()
```

```python
#healthy people fall below the average number of Calories consumed
Average_Calories = drop_data['Calories'].mean()
healthy_people = drop_data[drop_data['Calories'] < Average_Calories]
print(healthy_people)
Average_Calories
```

```python
import seaborn as sns
#checking how time spent affects calories burned
sns.regplot(x = "Duration", y = "Calories", data = drop_data)
```

```python
#correlation between Duration and calories
drop_data[["Duration", "Calories"]].corr()
```

```python
#checking how avg heart rate affects calories burned
sns.regplot(x = "Pulse", y = "Calories", data = drop_data)
```

```python
#correlation between Pulse and Calories 
drop_data[["Pulse", "Calories"]].corr()
```

```python
#checking how max heart rate affects calories burned
sns.regplot(x = "Maxpulse", y = "Calories", data = drop_data)
```

```python
#correlation between Maxpulse and Calories 
drop_data[["Maxpulse", "Calories"]].corr()
```

```python
drop_data["Calories_Categories"] = pd.cut(drop_data["Calories"],
                                          bins=3,  # Create 3 equal-width bins
                                          labels=["Not healthy", "Average", "Healthy"])
drop_data
```

```python
pie1=drop_data.groupby("Calories_Categories")["Calories"].count()
pie1.plot.pie(autopct="%1.1f%%", startangle=40, shadow=True, title="Calories Categories", ylabel="", figsize=(6, 6))
```
